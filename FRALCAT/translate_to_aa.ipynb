```python
!pip install biopython

from google.colab import drive
import csv
from Bio import SeqIO
from Bio.SeqRecord import SeqRecord

# Mount Google Drive
drive.mount('/content/drive')

# Define the input and output file paths
input_fasta = "/content/drive/MyDrive/GateLab-ProjetoTT5- Marie Anne/Dados/Dados/contigs_remontados/dados-contigs-remontados/Transcriptoma_remontagem/Transcriptoma_remontagem/Resultados_Egg_NOG/Dados-subconjuntos-diagrama-Venn/Grupo11/grupo11.fasta"
output_fasta = "/content/drive/MyDrive/GateLab-ProjetoTT5- Marie Anne/Dados/Dados/contigs_remontados/dados-contigs-remontados/Transcriptoma_remontagem/Transcriptoma_remontagem/Resultados_Egg_NOG/Dados-subconjuntos-diagrama-Venn/Grupo11/grupo11-amino.fasta"
output_csv = "/content/drive/MyDrive/GateLab-ProjetoTT5- Marie Anne/Dados/Dados/contigs_remontados/dados-contigs-remontados/Transcriptoma_remontagem/Transcriptoma_remontagem/Resultados_Egg_NOG/Dados-subconjuntos-diagrama-Venn/Grupo11/grupo11_lengths.csv"

# Read the sequences from the input multi-FASTA file
nucleotide_sequences = SeqIO.parse(input_fasta, "fasta")

# Function to check if the sequence has a valid stop codon at the end
def is_valid_translation(nucleotide_seq):
    return len(nucleotide_seq) % 3 == 0 and nucleotide_seq[-3:].translate() == "*"

# List to store the protein sequences and their lengths
protein_sequences = []
sequence_lengths = []

# Translate each nucleotide sequence to a protein sequence
for nucleotide_seq in nucleotide_sequences:
    try:
        protein_seq = nucleotide_seq.seq.translate(to_stop=True)
        valid_translation = is_valid_translation(nucleotide_seq.seq)
        status = "translated correctly" if valid_translation else "possible truncation or frame shift"
        description = f"{len(protein_seq)} amino acids, {status}"

        # Create a SeqRecord for each protein sequence
        protein_record = SeqRecord(protein_seq, id=nucleotide_seq.id, description=description)
        protein_sequences.append(protein_record)

        # Append sequence length and header to the list for CSV
        sequence_lengths.append((nucleotide_seq.id, len(protein_seq)))

    except Exception as e:
        print(f"Error translating sequence {nucleotide_seq.id}: {e}")

# Sort the protein sequences and sequence lengths by sequence length in ascending order
protein_sequences.sort(key=lambda x: len(x.seq))
sequence_lengths.sort(key=lambda x: x[1])

# Write the sorted protein sequences to the output FASTA file
SeqIO.write(protein_sequences, output_fasta, "fasta")
print(f"Translated and sorted sequences have been saved to {output_fasta}")

# Write the sequence headers and lengths to the CSV file
with open(output_csv, 'w', newline='') as csvfile:
    csvwriter = csv.writer(csvfile)
    csvwriter.writerow(["Header", "Length"])  # CSV header
    csvwriter.writerows(sequence_lengths)     # Write each sequence's header and length

print(f"CSV file with sequence lengths saved to {output_csv}")
```
