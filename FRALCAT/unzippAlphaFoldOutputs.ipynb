```python
# Step 2: Import necessary libraries
import os
import zipfile

# Step 3: Define the path to the directory containing the zipped folders
drive_directory = '/content/drive/MyDrive/tematico_ss/ptns_hipoteticas'  # Change 'your_directory' to your specific directory

# Step 4: Function to extract all zip files in the directory
def extract_all_zips(directory):
    # List all files in the directory
    for filename in os.listdir(directory):
        # Check if the file is a zip file
        if filename.endswith('.zip'):
            # Define the full path to the zip file
            zip_path = os.path.join(directory, filename)
            # Define the extraction path, maintaining the zip file name
            extract_path = os.path.join(directory, filename[:-4])  # Removing '.zip' from the folder name
            # Create the extraction directory if it doesn't exist
            os.makedirs(extract_path, exist_ok=True)
            # Extract the zip file
            with zipfile.ZipFile(zip_path, 'r') as zip_ref:
                zip_ref.extractall(extract_path)
            print(f'Extracted {filename} to {extract_path}')

# Step 5: Extract all zipped folders in the specified directory
extract_all_zips(drive_directory)
```
